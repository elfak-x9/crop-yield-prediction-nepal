```python
# 2. API Configuration (Using required lowercase keys for API compliance)
BASE_URL = "https://power.larc.nasa.gov/api/temporal/monthly/point"
START_YEAR = "1981"
END_YEAR = "2024"

# CRITICAL FIX: Parameter string converted to LOWERCASE
PARAMETERS = "prectotcorr,rh2m,t2m,t2m_max,t2m_min,ws10m"

all_data = []

print("Starting data extraction from NASA POWER API...")
```

```python
# 3. Request Data Loop
for province_name, coords in DISTRICTS.items():
    print(f"Fetching data for {province_name} District...")

    payload = {
        "parameters": PARAMETERS,
        "community": "AG",
        "longitude": coords["lon"],
        "latitude": coords["lat"],
        "start": START_YEAR,
        "end": END_YEAR,
        "format": "JSON",
    }

    try:
        response = requests.get(BASE_URL, params=payload)
        
        # If server rejects, print text response instead of crashing on json decode
        if response.status_code != 200:
            print(f"❌ Server Error {response.status_code} for {province_name}: {response.text}")
            continue
            
        data = response.json()
        records = data["properties"]["parameter"]

        # Aligns data rows based on returned keys (API returns dict keys in uppercase)
        for date_key in records["T2M"].keys():
            if date_key.endswith("13") or "ANN" in date_key:
                continue

            formatted_date = f"{date_key[:4]}-{date_key[4:]}"

            row = {
                "DATE": formatted_date,
                "PROVINCE": province_name,
                "LAT": coords["lat"],
                "LON": coords["lon"],
                "PRECTOT": records["PRECTOTCORR"].get(date_key),  
                "RH2M": records["RH2M"].get(date_key),
                "T2M": records["T2M"].get(date_key),
                "T2M_MAX": records["T2M_MAX"].get(date_key),
                "T2M_MIN": records["T2M_MIN"].get(date_key),
                "WS10M": records["WS10M"].get(date_key),
            }
            all_data.append(row)

        time.sleep(1)  # Rate limiting courtesy pause

    except Exception as e:
        print(f"❌ System Error for {province_name}: {e}")
```

```python
# 4. Save compilation to CSV file
if all_data:
    df = pd.DataFrame(all_data)
    column_order = ["DATE", "PROVINCE", "LAT", "LON", "PRECTOT", "RH2M", "T2M", "T2M_MAX", "T2M_MIN", "WS10M"]
    df = df[column_order]

    output_filename = "nepal_province_climate_1979_2024.csv"
    df.to_csv(output_filename, index=False)
    print(f"\n✅ Success! Dataset saved to: {output_filename}")
    print(df.head(5))
else:
    print("\n❌ Process failed. No data compiled.")
```
